```python
def FizzBuzz(n):
    output_list = []
    for i in n:
        output = ""
        if i % 3 == 0:
            output += "Fizz"
        if i % 5 == 0:
            output += "Buzz"
        if i % 7 == 0:
            output += "Fang"
        if i % 11 == 0:
            output += "Bang"
        output_list.append(output if output else i)
    return output_list
```

```python
def FizzBuzz(n):
    output_list = []
    for i in n:
        output = ""
        if i % 3 == 0:
            output += "Fizz"
        if i % 5 == 0:
            output += "Buzz"
        if i % 7 == 0:
            output += "Fang"
        if i % 11 == 0:
            output += "Bang"
        if output:
            output_list.append(output)
        else:
            output_list.append(i)
    return output_list
```

```python
my_list = []
for i in range(1,101):
    my_list.append(i)
result = FizzBuzz(my_list)

for index, value in enumerate(result, start=1):
    print(f"{index}. {value}")
```

```python
## CUSTOM VERSION

def fizz_buzz(numbers, rules):
    """
    numbers: an iterable of ints
    rules: list of (divisor, word) tuples, e.g. [(3, "Fizz"), (5, "Buzz")]
    """
    output_list = []
    for i in numbers:
        output = ""
        for divisor, word in rules:
            if i % divisor == 0:
                output += word
        output_list.append(output if output else str(i))
    return output_list


def print_results(result):
    for index, value in enumerate(result, start=1):
        print(f"{index}. {value}")
```
